Создайте массив из 20 случайных чисел и найдите их сумму с накоплением.

Функция numpy.cumsum возвращает массив с накопленной суммой элементов по заданному измерению

```python
asdrandom = np.random.randint(1,100, size = 20)
print(asdrandom)

cumsum = np.cumsum(asdrandom)
cumsum
```

Создайте матрицу 4×4 и найдите её собственные значения и векторы.

numpy.linalg.eig() вычисляет собственные значения и правые собственные векторы квадратной матрицы.

```python
my_matrix = np.random.randint(1,100, size = (4,4))
my_matrix

a,b = np.linalg.eig(my_matrix)
print("значения:")
print(a)

print("Векторы:")
print(b)
```

Создайте массив 6×6 и замените элементы главной диагонали на 0.

Функция numpy.diag_indices(n, ndim=2) возвращает кортеж массивов индексов для доступа к главной диагонали квадратного или многомерного массива.

```python
a = np.arange(36)
b = a.reshape(6,6)
c = np.diag_indices(6)
b[c] = 0
b
```
