Примените функцию np.select для категоризации массива по условиям.


```python

arr = np.array([5, 15, 25, 35, 45])

conditions = [arr < 10, (arr >= 10) & (arr < 30), arr >= 30]
choices = ["Низкий", "Средний", "Высокий"]

result = np.select(conditions, choices, default="Неизвестно")
print(result)
```

Реализуйте свёртку (convolution) двух одномерных массивов.



```python

a = np.array([1, 2, 3])
b = np.array([0, 1, 0.5])

result = np.convolve(a, b, mode="full")
print(result)
```

```python
arr1 = np.array([0, 0, 0], dtype=int)

arr2 = np.linspace(0.1, 0.3, len(arr1), dtype=float)

arr3 = np.random.logistic(loc=0.0, scale=1.0, size=len(arr1)).astype(float)

res_sum = ((arr1 + arr2 + arr3) * 5).astype(float)
res_sub = (arr1 - arr2).astype(float)
res_mul = (arr1 * arr3).astype(float)
res_div = (arr1 / arr2).astype(float)
res_pow = (arr1 ** 3).astype(float)

print("Массив 1:", arr1)
print("Массив 2:", arr2)
print("Массив 3:", arr3)
print("Сумма * 5:", res_sum)
print("Разность:", res_sub)
print("Умножение:", res_mul)
print("Деление:", res_div)
print("Степень 3:", res_pow)
```

```python

data = np.array([
    [65.20, 64.80, 66.10, 65.50],
    [65.60, 65.10, 66.40, 66.00],
    [65.90, 65.00, 66.00, 65.20],
    [65.10, 64.50, 65.40, 64.80],
    [64.90, 64.60, 65.50, 65.30]
], dtype=float)

mean_prices = np.mean(data, axis=1).reshape(-1, 1)

merged_data = np.hstack((data, mean_prices))

min_val = np.min(merged_data, axis=0)
max_val = np.max(merged_data, axis=0)
mean_val = np.mean(merged_data, axis=0)
variance = np.var(merged_data, axis=0)
std_dev = np.std(merged_data, axis=0)
root_mean_sq = np.sqrt(np.mean(merged_data**2, axis=0))
cv = std_dev / mean_val

print("Минимум:", min_val)
print("Максимум:", max_val)
print("Мат. ожидание:", mean_val)
print("Дисперсия:", variance)
print("Станд. отклонение:", std_dev)
print("Среднеквадратичное:", root_mean_sq)
print("Коэф. вариации:", cv)

np.save("alrosa_stats.npy", merged_data)
```
